## **Filterisasi**

```python
import pandas as pd

# === 1️⃣ Baca file langsung dari Google Drive ===
file_name = "/content/drive/MyDrive/TUGAS7PPW/webuage.csv"
print(f"📂 Membaca file: {file_name}")

# === 2️⃣ Baca file dengan delimiter otomatis ===
df = pd.read_csv(file_name, sep=None, engine="python")

# === 3️⃣ Normalisasi nama kolom agar mudah diakses ===
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# === 4️⃣ Konversi waktu ke datetime ===
df["request_time"] = pd.to_datetime(df["request_time"], errors="coerce")

# === 5️⃣ Filter hanya method GET dan halaman index/faq ===
filtered = df[
    (df["request_method"] == "GET") &
    (df["request_uri"].isin(["/faq.html", "/index.html"])) &
    (df["status"] == 200)
].copy()

# === 6️⃣ Ganti halaman menjadi A/B ===
filtered["url"] = filtered["request_uri"].replace({
    "/index.html": "A",
    "/faq.html": "B"
})

# === 7️⃣ Urutkan berdasarkan IP dan waktu ===
filtered = filtered.sort_values(by=["remote_host", "request_time"]).reset_index(drop=True)

# === 8️⃣ Hitung selisih waktu antarakses (per IP) ===
filtered["selisih_menit"] = (
    filtered.groupby("remote_host")["request_time"]
    .diff()
    .dt.total_seconds()
    .div(60)
    .fillna(0)
    .round(2)
)

# === 9️⃣ Buat kolom sesi per IP (gap ≥ 30 menit = sesi baru) ===
session_timeout = 30  # menit
filtered["sesi"] = (
    filtered.groupby("remote_host")["selisih_menit"]
    .transform(lambda x: (x > session_timeout).cumsum() + 1)
)

# === 🔟 Pilih kolom utama ===
final_table = filtered[[
    "remote_host", "request_time", "url", "status",
    "size_of_response_(incl._headers)", "selisih_menit", "sesi"
]].rename(columns={"remote_host": "ip"})

print("📋 Semua data ditampilkan dengan pembagian sesi per 30 menit:")
display(final_table.head(20))

# === 11️⃣ Simpan hasil ke file CSV di Google Drive ===
output_file = "/content/drive/MyDrive/TUGAS7PPW/sessionized_full_log_fixed.csv"
final_table.to_csv(output_file, index=False)

print(f"✅ File berhasil disimpan di:\n{output_file}")
```

```python
import pandas as pd
from google.colab import files

# Pastikan final_table sudah berisi ip, request_time, url, selisih_menit
df = final_table.copy()

# === 1️⃣ Tentukan batas sesi (30 menit) ===
session_timeout = pd.Timedelta(minutes=30)

# === 2️⃣ Urutkan berdasarkan IP dan waktu ===
df = df.sort_values(by=["ip", "request_time"]).reset_index(drop=True)

# === 3️⃣ Buat kolom sesi berdasarkan gap waktu antar akses per IP ===
df["sesi"] = (
    df.groupby("ip")["request_time"]
    .diff()
    .gt(session_timeout)
    .groupby(df["ip"])
    .cumsum()
    .add(1)
)

# === 4️⃣ Tandai halaman A dan B dalam setiap sesi ===
df["A"] = (df["url"] == "A").astype(int)
df["B"] = (df["url"] == "B").astype(int)

# === 5️⃣ Rekap per IP dan sesi (gabungkan kunjungan dalam satu periode)
result = (
    df.groupby(["ip", "sesi"], as_index=False)[["A", "B"]]
    .max()  # jika pernah ke A atau B dalam sesi tersebut = 1
)

# === 6️⃣ Hapus kolom sesi agar tampil seperti log ringkas ===
result = result.drop(columns=["sesi"])

# === 7️⃣ Tampilkan hasil akhir ===
print("📋 Ringkasan kunjungan per IP (setiap gap ≥ 30 menit buat baris baru):")
display(result.head(20))

# === 8️⃣ Simpan ke CSV ===
result.to_csv("ip_page_session_30min.csv", index=False)
files.download("ip_page_session_30min.csv")
```
